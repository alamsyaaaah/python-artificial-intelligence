```python
import numpy as np

# Data pelemparan dadu 25 kali
data = [1, 2, 3, 4, 5, 5, 6, 2, 3, 4, 5, 6, 6, 4, 3, 2, 1, 4, 3, 5, 6, 6, 5, 4, 5]

# Mencari nilai unik dan jumlah kemunculannya
nilai, frekuensi = np.unique(data, return_counts=True)

# Mencari indeks dengan frekuensi terbesar
indeks = np.argmax(frekuensi)

# Menentukan modus
modus = nilai[indeks]

print("Modus =", modus)
print("Frekuensi =", frekuensi[indeks])
```

```python
import numpy as np

# Data
data = [9, 10, 11, 6, 8, 7, 7, 5, 4, 5]

# Menghitung persentil
P25 = np.percentile(data, 25)
P75 = np.percentile(data, 75)

print("Persentil ke-25 =", P25)
print("Persentil ke-75 =", P75)
```

```python
import numpy as np

# Data nilai ujian
data = [50, 54, 62, 50, 52, 59, 61, 63, 65, 10, 53, 63, 65, 50, 59, 62, 50, 51, 57, 60, 63, 65, 65, 53, 99]

# a. Mengurutkan data
data_sorted = np.sort(data)

# b. Menghitung Q1, Q3, dan IQR
Q1 = np.percentile(data_sorted, 25)
Q3 = np.percentile(data_sorted, 75)
IQR = Q3 - Q1

# c. Menghitung lower bound dan upper bound
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

# d. Menentukan outlier
outlier = data_sorted[
    (data_sorted < lower_bound) | (data_sorted > upper_bound)
]

print("Data setelah diurutkan:")
print(data_sorted)

print("\nQ1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)

print("\nLower Bound =", lower_bound)
print("Upper Bound =", upper_bound)

print("\nOutlier =", outlier)
```

```python
import numpy as np

# Interval berat badan dan frekuensi
interval = [
    (41, 45),
    (46, 50),
    (50, 55),
    (56, 60),
    (61, 65),
    (66, 70)
]

frekuensi = [2, 3, 6, 4, 5, 2]

# Menghitung nilai tengah setiap interval
nilai_tengah = [(bawah + atas) / 2 for bawah, atas in interval]

print("Nilai tengah setiap interval:")
print(nilai_tengah)

# Membuat array data berdasarkan frekuensi
data_berat = np.repeat(nilai_tengah, frekuensi)

print("\nArray data berat badan:")
print(data_berat)

# Menghitung standard deviation
standar_deviasi = np.std(data_berat)

print("\nStandard Deviation =", standar_deviasi)
```
